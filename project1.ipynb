```python
# Question 2 b)
codon_table = { "UUU":"F", "UUC":"F", "UUA":"L", "UUG":"L", "CUU":"L", "CUC":"L", "CUA":"L", "CUG":"L", "AUU":"I", "AUC":"I", "AUA":"I", "AUG":"M", "GUU":"V", "GUC":"V", "GUA":"V", "GUG":"V", "UCU": "S", "UCC":"S", "UCA":"S", "UCG":"S","CCU":"P","CCC":"P", "CCA":"P", "CCG":"P", "ACU": "T", "ACC":"T", "ACA":"T","ACG":"T", "GCU":"A","GCC":"A", "GCA":"A", "GCG":"A", "UAU":"Y", "UAC":"Y", "UAA":"*", "UAG":"*", "CAU":"H", "CAC":"H", "CAA":"Q", "CAG":"Q", "AAU":"N","AAC":"N", "AAA":"K","AAG":"K", "GAU":"D","GAC":"D","GAA":"E","GAG":"E", "UGU":"C","UGC":"C","UGA":"*","UGG":"W","CGU":"R","CGC":"R","CGA":"R","CGG":"R","AGU":"S","AGC":"S","AGA":"R","AGG":"R","GGU":"G","GGC":"G","GGA":"G","GGG":"G"}
protein= ""
cds = str(cds).replace("T", "U")
for i in range (0,len(cds),3):
    codon = str(cds[i:i+3])
    amino_acid = codon_table[codon]
    protein += amino_acid

print("First 30 residues:", protein[:30])
print("Total protein length:", len(protein))
```

```python
# Question 4 a)
from Bio import SeqIO
genes = [
    ["MSRB","A","MSRB_gene.fasta","MSRB_ptn.fasta",44,394],
    ["MT-ND1", "B","NT_ND1_gene.fasta","NT_ND1_Ptn.fasta", 3307, 4262],
    ["ACTB", "C", "ACTB_gene.fasta", "ACTB_ptn.fasta", 85, 1212]
]

print("Gene | Category | CDS Length | Translated Length | Deposited Length | Identical | First Mismatch")

for gene in genes:

    name = gene[0]
    category = gene[1]
    gene_file = gene[2]
    protein_file = gene[3]
    start = gene[4]
    end = gene[5]

    sequence = SeqIO.read(gene_file, "fasta").seq

    cds = sequence[start-1:end]
    cds_for_translation = cds[:len(cds) - (len(cds) % 3)]
    translated = str(cds_for_translation.translate())
    if translated.endswith("*"):
        translated = translated[:-1]

    deposited = str(SeqIO.read(protein_file, "fasta").seq)

    if translated == deposited:
        identical = "Yes"
        mismatch = "No"

    else:
        identical = "No"

    for i, (a, b) in enumerate(zip(translated, deposited)):
        if a != b:
            mismatch = i + 1
            break

    print(name, category, len(cds), len(translated),
          len(deposited), identical, mismatch, sep=" | ")
```
