```python
# Question 2 a)
from Bio import SeqIO
sequence = SeqIO.read("ACTB_gene.fasta","fasta")
cds = sequence.seq[84:1212]
print("CDS length:",len(cds))
print(len(cds)%3)
```

```python
# Question 2 b)
codon_table = { "UUU":"F", "UUC":"F", "UUA":"L", "UUG":"L", "CUU":"L", "CUC":"L", "CUA":"L", "CUG":"L", "AUU":"I", "AUC":"I", "AUA":"I", "AUG":"M", "GUU":"V", "GUC":"V", "GUA":"V", "GUG":"V", "UCU": "S", "UCC":"S", "UCA":"S", "UCG":"S","CCU":"P","CCC":"P", "CCA":"P", "CCG":"P", "ACU": "T", "ACC":"T", "ACA":"T","ACG":"T", "GCU":"A","GCC":"A", "GCA":"A", "GCG":"A", "UAU":"Y", "UAC":"Y", "UAA":"*", "UAG":"*", "CAU":"H", "CAC":"H", "CAA":"Q", "CAG":"Q", "AAU":"N","AAC":"N", "AAA":"K","AAG":"K", "GAU":"D","GAC":"D","GAA":"E","GAG":"E", "UGU":"C","UGC":"C","UGA":"*","UGG":"W","CGU":"R","CGC":"R","CGA":"R","CGG":"R","AGU":"S","AGC":"S","AGA":"R","AGG":"R","GGU":"G","GGC":"G","GGA":"G","GGG":"G"}
protein= ""
cds = str(cds).replace("T", "U")
for i in range (0,len(cds),3):
    codon = str(cds[i:i+3])
    amino_acid = codon_table[codon]
    protein += amino_acid

print("First 30 residues:", protein[:30])
print("Total protein length:", len(protein))
```

```python
# Question 2 c)
deposited = SeqIO.read("ACTB_ptn.fasta", "fasta")
deposited_protein = str(deposited.seq)

if protein[0:30] == deposited_protein[0:30]:
    print("The two sequences are identical.")
else:
    print("The two sequences are not identical.")
```

```python
# Question 2 d) 
from Bio.Seq import Seq
Translated_ptn = str(Seq(cds).translate())
if protein[0:30] == Translated_ptn[0:30]:
    print("These are matching")
else:
    print("These are not matching")
```
